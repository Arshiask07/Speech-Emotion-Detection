```python
import os

def count_wavs(path):
    total = 0
    for root, dirs, files in os.walk(path):
        total += sum(1 for f in files if f.endswith('.wav'))
    return total

print("RAVDESS files:", count_wavs('data/RAVDESS'))
print("CREMA-D files:", count_wavs('data/CREMA-D'))
```
